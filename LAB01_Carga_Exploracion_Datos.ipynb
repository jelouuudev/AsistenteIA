## Punto 6 - Actividad de comprobación

```python
# 1. ¿Cuántos registros contiene el dataset?
print("1. Registros:", df.shape[0])
```

```python
# 2. ¿Cuántas columnas contiene?
print("2. Columnas:", df.shape[1])
```

```python
# 3. ¿Cuáles son los nombres de las columnas?
print("3. Nombres de columnas:", list(df.columns))
```

```python
# 4. ¿Qué tipo de dato tiene la columna Cantidad?
print("4. Tipo de dato de Cantidad:", df["Cantidad"].dtype)
```

```python
# 5. ¿Qué tipo de dato tiene inicialmente la columna Fecha?
print("5. Tipo de dato de Fecha:", df["Fecha"].dtype)
```

```python
# 6. Según describe(), ¿cuál es el precio máximo registrado?
stats = df.describe()
print("6. Precio máximo registrado:", stats.loc["max", "Precio"])
```

```python
# 7. Según describe(), ¿cuál es la cantidad máxima registrada en una venta?
print("7. Cantidad máxima registrada:", stats.loc["max", "Cantidad"])
```
